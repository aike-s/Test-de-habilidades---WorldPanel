# Primer test de habilidades - WorldPanel

A continuación la resolucion del primer test enviado incialmente como un archivo comprimido.


```python
import pandas as pd
```

1. Leer el archivo ventas.csv y mostrar los primeros 5 registros.

```python
df_ventas = pd.read_csv("test_ventas.csv")
df_productos = pd.read_csv("test_productos.csv")
```

```python
df_ventas.head(5)
```

2. Calcular el monto total de cada venta (cantidad * precio_unitario) y agregarlo como nueva
columna.

```python
df_ventas['monto_total'] = df_ventas.cantidad * df_ventas.precio_unitario

df_ventas.head(5)
```

3. Obtener la suma total de ventas.

```python
print(df_ventas.monto_total.sum())
```

4. Calcular el promedio de monto vendido por cliente.

```python
client_grouped = df_ventas.groupby('id_cliente')
result = client_grouped.monto_total.agg('mean')

print(result)
```

5. Filtrar las ventas donde el monto total sea mayor a $200.

```python
result = df_ventas[df_ventas['monto_total'] > 200]

print(result)
```

6. Agrupar por fecha y obtener el total vendido por día

```python
date_grouped = df_ventas.groupby('fecha')
result = date_grouped.monto_total.agg('sum')

print(result)
```

7. Unir los archivos ventas.csv y productos.csv por id_producto.


```python
df_merge_vp = pd.merge(df_ventas, df_productos, on='id_producto')

df_merge_vp.head(5)
```

8. Agrupar por producto y obtener cuántas unidades se vendieron de cada uno

```python
product_grouped = df_merge_vp.groupby('id_producto')['cantidad'].sum()

product_grouped.head(5)
```

9. Obtener el total de ventas por categoría de producto.

```python
df_merge_vp['monto_total'] = df_merge_vp.cantidad * df_merge_vp.precio_unitario
category_grouped = df_merge_vp.groupby('categoria')['monto_total'].sum()

category_grouped.head(5)
```

10. Listar los 3 productos más vendidos (por cantidad).

```python
sum_category_grouped = df_merge_vp.groupby('nombre_producto').agg(cantidad_total=('cantidad', 'sum'))
sum_category_grouped['rank'] = sum_category_grouped['cantidad_total'].rank(ascending=False)
sum_category_grouped = sum_category_grouped.sort_values(by='rank', ascending=True)

sum_category_grouped.head(3)
```
